```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Step 1: Data
X = np.array([[1], [2], [3], [4]]) # Study hours
y = np.array([20, 40, 60, 80]) # Marks
```

```python
# Step 2: Create and train model
model = LinearRegression()
model.fit(X, y)
```

```python
# Step 3: Get slope (m) and intercept (c)
m = model.coef_[0]
c = model.intercept_
print(f"Equation: Y = {m:.2f}X + {c:.2f}")
```

```python
# Step 4: Predict marks for 5 hours
prediction = model.predict([[5]])
print(f"Predicted marks for 5 study hours:{prediction[0]:.2f}")
```

```python
# Step 5: Plot data and regression line
plt.scatter(X, y, color='blue', label="Actual data")
plt.plot(X, model.predict(X), color='red', label=("Regression line"))
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.show()
```
